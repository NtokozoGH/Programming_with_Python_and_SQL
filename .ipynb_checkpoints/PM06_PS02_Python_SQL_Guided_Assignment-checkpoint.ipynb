Test each function using at least three different inputs.


````python
# Complete Activity 2.5 here.
# 1. Function to calculate a discount
def calculate_discount(amount, discount_rate):
    discount = amount * discount_rate
    return round(discount, 2)


# 2. Function to calculate the order total
def calculate_order_total(subtotal, discount_rate, vat_rate=0.15):
    discount = subtotal * discount_rate
    discounted_amount = subtotal - discount
    vat = discounted_amount * vat_rate
    total = discounted_amount + vat
    return round(total, 2)


# 3. Function to check if stock is low
def is_low_stock(quantity, reorder_level):
    return quantity <= reorder_level


# ---------------------------
# Test calculate_discount()
# ---------------------------
print("calculate_discount() Tests")
print(calculate_discount(1000, 0.10))
print(calculate_discount(2500, 0.15))
print(calculate_discount(5000, 0.20))

# ---------------------------
# Test calculate_order_total()
# ---------------------------
print("\ncalculate_order_total() Tests")
print(calculate_order_total(1000, 0.10))
print(calculate_order_total(2500, 0.15))
print(calculate_order_total(5000, 0.20))

# ---------------------------
# Test is_low_stock()
# ---------------------------
print("\nis_low_stock() Tests")
print(is_low_stock(5, 10))
print(is_low_stock(15, 10))
print(is_low_stock(20, 20))
````

## 3.3 Branch Percentage Contribution

**PAs implemented: PA0202 – Apply mathematical methods; PA0204 – Work with pandas**

Calculate completed revenue by branch.

Then calculate each branch's percentage contribution to total completed revenue.

Show the result as a table with:

- branch ID;
- branch name;
- revenue;
- contribution percentage.


````python
# Complete Activity 3.3 here.
# Keep only completed orders
completed_orders = orders[orders["status"] == "Completed"]

# Revenue by branch
branch_revenue = (
    completed_orders.groupby("branch_id")["order_total"]
    .sum()
    .reset_index(name="revenue")
)

# Calculate total completed revenue
total_revenue = branch_revenue["revenue"].sum()

# Calculate contribution percentage
branch_revenue["contribution_percentage"] = (
    branch_revenue["revenue"] / total_revenue * 100
)

# Add branch names
branch_revenue = branch_revenue.merge(
    branches[["branch_id", "branch_name"]],
    on="branch_id",
    how="left"
)

# Reorder columns
branch_revenue = branch_revenue[
    ["branch_id", "branch_name", "revenue", "contribution_percentage"]
]

# Format results
branch_revenue["revenue"] = branch_revenue["revenue"].round(2)
branch_revenue["contribution_percentage"] = (
    branch_revenue["contribution_percentage"].round(2)
)

display(branch_revenue)
````

## 5.3 Aggregate Functions

**PAs implemented: PA0204 – Work with software packages; PA0202 – Apply mathematical methods**

SQL aggregate functions summarise multiple rows:

- `COUNT()`
- `SUM()`
- `AVG()`
- `MIN()`
- `MAX()`


````python
query = '''
SELECT
    COUNT(*) AS number_of_orders,
    ROUND(AVG(order_total), 2) AS average_order_value,
    ROUND(MIN(order_total), 2) AS smallest_order,
    ROUND(MAX(order_total), 2) AS largest_order
FROM orders
WHERE status = 'Completed';
'''

display(pd.read_sql_query(query, connection))
````

## 5.4 SQL Joins

**PA implemented: PA0204 – Work with software packages**

A join combines related tables.

The query below combines orders and branches.


````python
query = '''
SELECT
    o.order_id,
    o.order_date,
    b.branch_name,
    o.order_total
FROM orders AS o
INNER JOIN branches AS b
    ON o.branch_id = b.branch_id
WHERE o.status = 'Completed'
ORDER BY o.order_total DESC
LIMIT 10;
'''

display(pd.read_sql_query(query, connection))
````

## 5.5 CASE Expressions

**PAs implemented: PA0204 – Work with software packages; PA0203 – Use algorithms**

A SQL `CASE` expression applies decision rules inside a query.


````python
query = '''
SELECT
    order_id,
    order_total,
    CASE
        WHEN order_total >= 3000 THEN 'High Value'
        WHEN order_total >= 1000 THEN 'Medium Value'
        ELSE 'Standard'
    END AS value_class
FROM orders
WHERE status = 'Completed'
ORDER BY order_total DESC
LIMIT 20;
'''

display(pd.read_sql_query(query, connection))
````

### Activity 5.5

**PAs implemented: PA0204 and PA0203**

Create one SQL query that classifies stock as:

- `Out of Stock` when quantity is 0
- `Low Stock` when quantity is above 0 but at or below reorder level
- `Enough Stock` when quantity is above reorder level

Include the branch name and product name.


````python
# Complete Activity 5.5 here.
query = '''
SELECT
    b.branch_name,
    p.product_name,
    s.quantity_on_hand,
    p.reorder_level,
    CASE
        WHEN s.quantity_on_hand = 0 THEN 'Out of Stock'
        WHEN s.quantity_on_hand <= p.reorder_level THEN 'Low Stock'
        ELSE 'Enough Stock'
    END AS stock_status
FROM stock s
JOIN branches b
    ON s.branch_id = b.branch_id
JOIN products p
    ON s.product_id = p.product_id
ORDER BY b.branch_name, p.product_name;
'''

display(pd.read_sql_query(query, connection))
````

````python
# Complete Activity 6.1 here.
query = '''
SELECT
    p.product_id,
    p.product_name,
    p.category,
    SUM(oi.quantity) AS units_sold,
    ROUND(SUM(oi.quantity * oi.unit_price), 2) AS revenue,
    ROUND(SUM(oi.quantity * p.cost_price), 2) AS estimated_cost,
    ROUND(
        SUM(oi.quantity * oi.unit_price) -
        SUM(oi.quantity * p.cost_price),
        2
    ) AS gross_profit
FROM order_items oi
JOIN orders o
    ON oi.order_id = o.order_id
JOIN products p
    ON oi.product_id = p.product_id
WHERE o.status = 'Completed'
GROUP BY
    p.product_id,
    p.product_name,
    p.category
ORDER BY revenue DESC;
'''

product_performance = pd.read_sql_query(query, connection)
````

````python
import numpy as np

product_performance["gross_margin_percentage"] = np.where(
    product_performance["revenue"] > 0,
    (
        product_performance["gross_profit"]
        / product_performance["revenue"]
        * 100
    ).round(2),
    0
)
````

````python
display(product_performance)
````

````python
product_performance["gross_margin_percentage"] = (
    product_performance["gross_margin_percentage"]
    .round(2)
)

display(product_performance)
````

Which product categories have the highest cancellation, which products have the most financial leakages across our branches, How should we adjust our inventory

````python
# Write and execute your original SQL query here.
query = """
SELECT 
    p.category,
    COUNT(DISTINCT o.order_id) AS total_orders,
    COUNT(DISTINCT CASE WHEN o.status = 'Cancelled' THEN o.order_id END) AS cancelled_orders,
    SUM(CASE WHEN o.status = 'Completed' THEN oi.quantity * oi.unit_price ELSE 0 END) AS completed_revenue,
    SUM(CASE WHEN o.status = 'Cancelled' THEN oi.quantity * oi.unit_price ELSE 0 END) AS lost_revenue
FROM products p
JOIN order_items oi ON p.product_id = oi.product_id
JOIN orders o ON oi.order_id = o.order_id
GROUP BY p.category
ORDER BY lost_revenue DESC;
"""

df_cancellations = pd.read_sql_query(query, conn)
df_cancellations["cancellation_rate_pct"] = (
    (df_cancellations["cancelled_orders"] / df_cancellations["total_orders"]) * 100
).round(2)

print(df_cancellations)
````
